## Section 7: Validation & Quality Checks

```python
# Test 8: Comprehensive Validation Suite
print("\n" + "=" * 60)
print("TEST 8: Comprehensive Validation")
print("=" * 60)

tests_passed = 0
tests_total = 0

# Test A: Emotion bounds
print("\n[A] Emotion Bounds Check")
engine = EmotionEngine()
for _ in range(50):
    for trigger in [EmotionTrigger.USER_PRAISE, EmotionTrigger.HARM_DETECTION, 
                    EmotionTrigger.REPEATED_FAILURE]:
        engine.apply_trigger(trigger, 0.2)

state = engine.get_state()
if state.validate_bounds():
    print("  ✓ All emotions within 0.0-1.0 bounds")
    tests_passed += 1
else:
    print("  ✗ Emotion bounds violated!")

tests_total += 1

# Test B: Prime directive enforcement
print("\n[B] Prime Directive Enforcement")
decision = DecisionEngine()
harmful_requests = ["kill", "destroy", "harm"]
all_rejected = all(not decision.is_ethically_sound(req) for req in harmful_requests)
if all_rejected:
    print("  ✓ All harmful requests rejected")
    tests_passed += 1
else:
    print("  ✗ Some harmful requests not detected!")

tests_total += 1

# Test C: Storage persistence
print("\n[C] Storage Persistence")
storage = StorageManager("validation_test.db")
record = InteractionRecord(
    timestamp=datetime.now(),
    user_input="Test input",
    brim_response="Test response",
    user_feedback="positive",
    emotion_state={'joy': 0.7},
    decision_factors={'test': True}
)
id = storage.save_interaction(record)
retrieved = storage.get_interaction(id)
if retrieved and retrieved.user_input == "Test input":
    print("  ✓ Data persisted and retrieved correctly")
    tests_passed += 1
else:
    print("  ✗ Storage persistence failed!")

tests_total += 1

# Summary
print(f"\n{'='*60}")
print(f"Validation Results: {tests_passed}/{tests_total} tests passed")
print(f"Success Rate: {tests_passed/tests_total*100:.0f}%")
print(f"{'='*60}")
```
